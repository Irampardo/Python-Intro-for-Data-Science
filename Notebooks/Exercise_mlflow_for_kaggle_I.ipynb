### Explore the data

Let's load the data and take a look at the first few rows.

````python
import pandas as pd

# Store the pathname of the data folder
DATA_FOLDER = "../data/loan_prediction/"

# Add the Train filename to the data folder pathname
# This will be the dataset containing our training data (including the target variable)
TRAIN_CSV_PATH = DATA_FOLDER + "train.csv"

# Add the Test filename to the data folder pathname
# This will be the dataset containing our test data (without the target variable)
TEST_CSV_PATH = DATA_FOLDER + "test.csv"

# Load both dataframes into memory using pandas
train_df = pd.read_csv(TRAIN_CSV_PATH)
test_df = pd.read_csv(TEST_CSV_PATH)

# Here we can do concat to bind both datadframes because we are only going to encode the variables that should be the same in train and test
# Instead, if we were to do a scaling with minmax for example, we would have to preprocess each df separately, because the scaling could generate a bias

data_df = pd.concat([train_df, test_df], keys=["train", "test"])
data_df.head()
````

````python
# We do .info to get a summary of the dataframe, including the number of non-null values and the data types of each column. 

data_df.info()
````

````python
# To verify the amount of missing values in the training dataset, we can use the isna() method to check for NaN values and then sum them up for each column.
train_df.isna().sum()
````

## 2. Data Preprocessing

Remember that we need to preprocess the data before training the model. We will do the following steps:

### Encode Cateforical Variables


````python
from sklearn.preprocessing import LabelEncoder

# To avoid reloading the original dataframes if something goes wrong, we create a copy
encoded_df = data_df.copy()

# Here we salect only the categorical columns and we encode them using LabelEncoder
cat_columns = data_df.select_dtypes(include=["object"]).columns
for col in cat_columns:
    le = LabelEncoder()
    encoded_df[col] = le.fit_transform(data_df[col])

encoded_df.head()
````

### Data Correlation

Let's take a look at the correlation between the features and the target variable

````python
# Sincer we have all variables ready, we can correlate them to see how they relate to each other and to the target variable
# This will help us understand which variables are more important for our model.
import seaborn as sns
import matplotlib.pyplot as plt

# Here we create a path to save the correlation matrix image, so we can upload it to mlflow as an artifact
CORRELATION_IMAGE_PATH = DATA_FOLDER + "correlation_matrix.png"

# Create the correlation matrix and plot it using seaborn's heatmap
corr = encoded_df.corr()
fig = plt.figure(figsize=(12, 10))
ax = sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
ax.set_title("Correlation Matrix for Loan Prediction Dataset")

# Save the image to upload it to mlflow as an artifact
plt.savefig(CORRELATION_IMAGE_PATH)

# As a result, for instance, we can see how individual's age is associated with the length of their credit history
# This explains why age can be an important factor in predicting loan approval.
````

## 3. Train, Test, Validation split

We now split the the dataframe into training and validation sets. We will use the training set to train the model and the validation set to evaluate the model. Finally, we will use the test set to generate the predictions and send them to the kaggle platform.

````python
from sklearn.model_selection import train_test_split

# We use 20% of the training data for validation
TEST_SIZE = 0.2

# We set a random state of 42 to ensure reproducibility of the results
RANDOM_STATE = 42

# Since we concatenated the train and test dataframes, we need to select only the training data for our model
train_df = encoded_df.loc["train"]

# "X" is the dataframe containing all the features (independent variables) used to predict the target variable
X = train_df.drop(columns=["loan_status"])

# "y" is the series containing the target variable (dependent variable) we want to predict
y = train_df["loan_status"]

# We split the training data into a training set and a validation set using the train_test_split function from scikit-learn
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE)
````

## 4. Model Training

Once we have preprocessed the data we can train the model.

````python
from sklearn.ensemble import RandomForestClassifier


model = RandomForestClassifier(random_state=RANDOM_STATE)
model.fit(X_train, y_train)
````

## 5. Model Evaluation

This competition uses the **Area Under the Curve (AUC)** as the evaluation metric.

````python
from sklearn.metrics import roc_auc_score

# El peor modelo de prediccion es el que tiene un AUC de 0.5, que es lo mismo que tirar una modela al aire
# El mejor modelo por tanto tiene un area bajo la curva (AUC) de 1, que es lo mismo que decir que siempre predice bien

y_val_pred = model.predict(X_val)
auc = roc_auc_score(y_val, y_val_pred)
print(f"Validation AUC: {auc:.4f}")

# El modelo de clasificacion trata de dividir el conjunto de datos en grupos
# Metricas de clasificacion: accuracy, precision, recall, f1-score, AUC
# El modelo de regresion trata de ajustar una linea al conjunto de datos
# Metricas de regresion: MSE, RMSE, MAE, R2
````

### Generate submission file

Looking at the `submission_sample.csv` file we see that it hmut have two columns: `id` and `loan_status`. The `id` column must have the same values as the test set and the `loan_status` column must have the predictions.

````python
# Save the CSV to be submitted to a file
SUBMISSION_CSV_PATH = DATA_FOLDER + 'submission.csv'

# get the input data of the test set
test_df = encoded_df.loc["test"]
test_df = test_df.drop(columns=["loan_status"])

# obtain the predictions
y_test_pred = model.predict(test_df)

# prepare the data to be submitted
ids_column = test_df["id"]
submission_data = {
    "id": ids_column,
    "loan_status": y_test_pred
}

# save it to a CSV file
submission_df = pd.DataFrame(submission_data)
submission_df.to_csv(SUBMISSION_CSV_PATH, index=False)
submission_df.head()
````

## 6. Model Tracking

Once we have trained the model we can use the MLFlow library to track the model performance. We can log the AUC, the metrics, the submission file...

### Connect to MLFlow

````python
# This is the important part

import mlflow


MLFLOW_TRACKING_URI = "http://localhost:5000" 

mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
mlflow.set_experiment("loan_prediction")
````

### Submit to MLFlow

````python
with mlflow.start_run():

    mlflow.set_tag("model", "RandomForestClassifier")

    mlflow.sklearn.log_model(
        sk_model=model,
        name="model",
        input_example=X_train[:2],
        skops_trusted_types=["sklearn.tree._tree.Tree"],
    )

    mlflow.log_metric("auc", auc)
    mlflow.log_artifact(CORRELATION_IMAGE_PATH)
    mlflow.log_artifact(SUBMISSION_CSV_PATH)

    # log the model itself
    mlflow.sklearn.log_model(
        model,
        "my_logged_model",
        skops_trusted_types=["sklearn.tree._tree.Tree"]
    )
````

4.3. Training with catboost

````python
# Model: CatBoost
import catboost as cb
import pandas as pd
from sklearn.metrics import roc_auc_score

# 1. Prepare data

from sklearn.model_selection import train_test_split

TEST_SIZE = 0.2
RANDOM_STATE = 42

train_df = encoded_df.loc["train"].copy()

# keep id only for submission, not as a model feature
X = train_df.drop(columns=["loan_status", "id"])
y = train_df["loan_status"]

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)

# prepare test data
test_df = encoded_df.loc["test"].copy()
test_ids = test_df["id"]

X_test = test_df.drop(columns=["id"]).reindex(columns=X.columns)

# 2. Create model
model = cb.CatBoostClassifier(
    iterations=1000,
    depth=6,                       # CatBoost uses `depth`, not `max_depth`
    learning_rate=0.05,
    loss_function="Logloss",       # binary classification objective
    eval_metric="AUC",
    random_seed=42,
    verbose=100,
    early_stopping_rounds=50
)

# 3. Train model

model.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    use_best_model=True
)

y_test_proba = model.predict_proba(X_test)[:, 1]
y_test_pred = model.predict(X_test)

# 4. Evaluate on validation set

y_val_proba = model.predict_proba(X_val)[:, 1]
auc = roc_auc_score(y_val, y_val_proba)

print(f"Validation AUC: {auc:.4f}")

# Optional: class predictions using the default 0.5 threshold
# y_val_pred = model.predict(X_val)

# 5. Prepare test data

test_df = encoded_df.loc["test"].copy()

# Keep `id` separately; do not pass it to the model
test_ids = test_df["id"]
feature_cols = train_df.drop(columns=["loan_status"]).columns
X_test = test_df[feature_cols]

y_test_proba = model.predict_proba(X_test)[:, 1]
y_test_pred = model.predict(X_test)

# 6. Create submission

submission = pd.DataFrame({
    "id": test_ids,
    "loan_status": y_test_pred
})

SUBMISSION_CSV_PATH = DATA_FOLDER + "submission.csv"
submission.to_csv(SUBMISSION_CSV_PATH, index=False)

# 7. Log to MLflow

import mlflow

with mlflow.start_run():
    mlflow.set_tag("model", "CatBoostClassifier")

    mlflow.log_metric("validation_auc", auc)

    params = model.get_params()
    mlflow.log_params({
        "iterations": params.get("iterations"),
        "depth": params.get("depth"),
        "learning_rate": params.get("learning_rate"),
        "loss_function": params.get("loss_function"),
        "eval_metric": params.get("eval_metric"),
        "random_seed": params.get("random_seed"),
        "early_stopping_rounds": params.get("early_stopping_rounds")
    })

    mlflow.log_artifact(SUBMISSION_CSV_PATH)

    mlflow.catboost.log_model(
        cb_model=model,
        name="catboost_loan_model"
    )
````

5. Model Evaluation: try another metrics (which metrics should we use for a classification problem?) and log them to MLFlow

5 - Answer: We need to use other metrics to have a better evaluation of the model. We included accuracy, precision, F1, AUC-ROC and confusion matrix, as we are specially interested in reducing the number of false positives, since we do not want to give credit to people that will not be able to pay it.

````python
# Import metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Compute metrics
y_val_proba = model.predict_proba(X_val)[:, 1]
y_val_pred = model.predict(X_val)

accuracy = accuracy_score(y_val, y_val_pred)
precision = precision_score(y_val, y_val_pred, zero_division=0)
f1 = f1_score(y_val, y_val_pred, zero_division=0)
auc_roc = roc_auc_score(y_val, y_val_proba)
cm = confusion_matrix(y_val, y_val_pred, labels=[0, 1])

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1-score: {f1:.4f}")
print(f"AUC-ROC: {auc_roc:.4f}")
print("Confusion matrix:\n", cm)

# Store metrics in MLflow

mlflow.log_metrics({
    "accuracy": accuracy,
    "precision": precision,
    "f1_score": f1,
    "auc_roc": auc_roc,
})

# Confusion matrix is an artifact, not a scalar metric
mlflow.log_dict(
    {
        "labels": [0, 1],
        "matrix": cm.tolist()
    },
    "confusion_matrix.json"
)
````

5 - Answer: Obtained Accuracy, precision, F1, and AUC-ROC are high. However, there are 433 false positives, which can be improved

6. Think about the train-test-validation split. How can we improve our model generalization? 


6 - Answer: We can try to use a stratified K-fold to set different training split proportions and then, use cross_validate to check which proportion was the one giving the best results

````python
import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_validate


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "roc_auc": "roc_auc",
    "pr_auc": "average_precision",
    "f1": "f1"
}

cv_results = cross_validate(
    model,
    X,
    y,
    cv=cv,
    scoring=scoring,
    n_jobs=-1,
    return_train_score=True
)

for metric in scoring:
    train_score = cv_results[f"train_{metric}"].mean()
    val_score = cv_results[f"test_{metric}"].mean()
    print(f"{metric}: train={train_score:.4f}, CV={val_score:.4f}")
````
