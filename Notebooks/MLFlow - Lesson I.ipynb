### MLFlow Lesson I

```python
import mlflow
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor

# Create the experiment if it doesn't exist
# mlflow.create_experiment("my_experiment", experiment_id="1")

#experiment_name = "diabetes_rf_experiment"

# Create if it doesn't exist (returns existing one if already present)
#experiment = mlflow.create_experiment(experiment_name)

# Now set it (by name or ID)
mlflow.set_tracking_uri("http://localhost:5000")  # Set the tracking URI to your MLflow server
#mlflow.set_experiment(experiment_name)
mlflow.set_experiment(experiment_id="1")
# or: mlflow.set_experiment(experiment_id=experiment)

mlflow.autolog()
```

Cargar los datos y crear un modelo sencillo

```python
db = load_diabetes()

X_train, X_test, y_train, y_test = train_test_split(db.data, db.target)

# Create and train models.
rf = RandomForestRegressor(n_estimators=100, max_depth=6, max_features=3)
rf.fit(X_train, y_train)

# Use the model to make predictions on the test dataset.
predictions = rf.predict(X_test)
```
